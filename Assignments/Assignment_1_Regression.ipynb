```python
def plot_distribution(values, title="Distribution"):
    plt.figure(figsize=(7, 4))
    sns.histplot(values, kde=True)
    plt.title(title)
    plt.show()

def plot_scatter(x, y, xlabel, ylabel, title=None):
    plt.figure(figsize=(6, 4))
    sns.scatterplot(x=x, y=y)
    plt.xlabel(xlabel)
    plt.ylabel(ylabel)
    if title:
        plt.title(title)
    plt.show()

def plot_correlation_heatmap(dataframe, title="Correlations"):
    plt.figure(figsize=(8, 6))
    sns.heatmap(dataframe.corr(numeric_only=True), annot=True, fmt=".2f", cmap="viridis")
    plt.title(title)
    plt.show()

def plot_fitted_line(x, y_true, y_pred, xlabel, ylabel, title="Fitted line"):
    x = np.asarray(x)
    order = np.argsort(x)
    plt.figure(figsize=(6, 4))
    plt.scatter(x, y_true, alpha=0.4)
    plt.plot(x[order], np.asarray(y_pred)[order], color="red")
    plt.xlabel(xlabel)
    plt.ylabel(ylabel)
    plt.title(title)
    plt.show()

def plot_predicted_vs_actual(y_true, y_pred, title):
    plt.figure(figsize=(6, 6))
    plt.scatter(y_true, y_pred, alpha=0.4)
    low = min(np.min(y_true), np.min(y_pred))
    high = max(np.max(y_true), np.max(y_pred))
    plt.plot([low, high], [low, high], color="red")
    plt.xlabel("Actual")
    plt.ylabel("Predicted")
    plt.title(title)
    plt.show()

def plot_residuals(y_pred, residuals, title="Residuals vs predicted"):
    plt.figure(figsize=(6, 4))
    plt.scatter(y_pred, residuals, alpha=0.4)
    plt.axhline(0, color="red")
    plt.xlabel("Predicted")
    plt.ylabel("Residual")
    plt.title(title)
    plt.show()
```
