### The Data

Run this cell to generate a noisy non-linear dataset and split it into train/test.

```python
X = np.sort(np.random.uniform(0, 1, 25)).reshape(-1, 1)
y = np.sin(2 * np.pi * X).ravel() + np.random.normal(0, 0.3, 25)

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

plt.figure(figsize=(8, 5))
plt.scatter(X_train, y_train, label='Train', s=40)
plt.scatter(X_test, y_test, label='Test', marker='s', s=40)
plt.xlabel('X')
plt.ylabel('y')
plt.title('Noisy Non-Linear Data')
plt.legend()
plt.tight_layout()
plt.show()
```
